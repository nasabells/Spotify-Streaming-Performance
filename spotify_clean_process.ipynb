```python
import pandas as pd
from google.colab import files
```

```python
df = pd.read_csv('Spotify Streaming Performance Dataset.csv')
```

```python
df.head(10)
```

```python
# Mengubah nama beberapa kolom sekaligus
df = df.rename(columns={
    'Artist Name': 'artist_name',
    'Country of Origin': 'country',
    'Primary Language': 'language',
    'Primary Genre': 'genre',
    'Artist Type': 'artist_type',
    'Debut Year': 'debut_year',
    'Total Streams (in millions)': 'total_streams',
    'Lead Streams (in millions)': 'lead_streams',
    'Feature Streams (in millions)': 'feature_streams',
    'Solo Streams (in millions)': 'solo_streams',
    '% of Solo Streams': 'percent_solo_streams',
    'Collaborative Streams (in millions)': 'collab_streams',
    '% of Collaborative Streams': 'percent_collab_streams',
    'Sex': 'sex'
})
```

```python
text_col = ['artist_name', 'country', 'language', 'genre', 'artist_type', 'sex']
for col in text_col:
  if col in df.columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
df['country'] = df['country'].fillna('Unknown')
df['genre'] = df['genre'].fillna('Other')
df['sex'] = df['sex'].fillna('Unknown')
df[['total_treams', 'lead_streams', 'feature_streams']] = df[['total_streams', 'lead_streams', 'feature_streams']].fillna(0)
```

```python
num_cols = ['total_streams', 'lead_streams', 'feature_streams', 'debut_year']
for col in num_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', ''), errors='coerce')
```

```python
df = df.sort_values(by='total_streams', ascending=False).drop_duplicates(subset=['artist_name'], keep='first')
```

```python
df = df[(df['total_streams'] >= 0) & (df['debut_year'] >= 1900)]
```

```python
df.to_csv('spotify_clean.csv', index=False)
files.download('spotify_clean.csv')
```
